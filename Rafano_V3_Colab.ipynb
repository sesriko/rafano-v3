## STEP 2: Setup Secrets
Klik ikon 🔑 **Secrets** di sidebar kiri Colab, tambahkan:
- `TELEGRAM_BOT_TOKEN` = token dari @BotFather
- `TARGET_CHAT_ID` = chat ID lu
- `ARJUM_API_KEY` = API key Arjum


```python
import os
from google.colab import userdata

def load_secrets():
    secrets = {}
    for key in ["TELEGRAM_BOT_TOKEN", "TARGET_CHAT_ID", "ARJUM_API_KEY"]:
        try:
            val = userdata.get(key)
            if val:
                os.environ[key] = val
                secrets[key] = f"✅ {len(val)} chars"
            else:
                secrets[key] = "❌ NOT SET"
        except Exception as e:
            secrets[key] = f"⚠️ {e}"
    return secrets

secrets = load_secrets()
for k,v in secrets.items():
    print(f"{k}: {v}")
```

## STEP 3: Download Code dari GitHub
Upload file ke GitHub dulu, atau upload manual ke Colab

```python
# GANTI USERNAME GITHUB LU
GITHUB_USERNAME = "USERNAME"
REPO_NAME = "rafano-v3"
BRANCH = "main"

import requests
def download_from_github():
    base_url = f"https://raw.githubusercontent.com/{GITHUB_USERNAME}/{REPO_NAME}/{BRANCH}/"
    files = ["rafano_v3.py", "rafano_db_free.py", "rafano_parallel_free.py"]
    for file in files:
        url = base_url + file
        try:
            r = requests.get(url, timeout=10)
            if r.status_code == 200:
                with open(f"/content/{file}", "w", encoding="utf-8") as f:
                    f.write(r.text)
                print(f"✅ Downloaded {file}")
            else:
                print(f"⚠️ Failed {file}: {r.status_code}")
        except Exception as e:
            print(f"❌ Error {file}: {e}")

if GITHUB_USERNAME != "USERNAME":
    download_from_github()
else:
    print("⚠️ Upload file manual ke Colab Files sidebar")
```
